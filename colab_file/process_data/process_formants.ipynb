```python
from google.colab import drive
drive.mount('/content/drive')

import os
import cv2
import numpy as np

BASE_PATH = "/content/drive/MyDrive/Thesis file/Formant"
OUTPUT_PATH = "/content/drive/MyDrive/Thesis file/Processed_Formant"

groups = {
    "HC": ["HC_F1_Images", "HC_F2_Images", "HC_F3_Images", "HC_F4_Images"],
    "PD": ["PD_F1_Images", "PD_F2_Images", "PD_F3_Images", "PD_F4_Images"]
}

def process_image(img_path):
    img = cv2.imread(img_path)
    if img is None:
        return None

    # 1. Resize
    img = cv2.resize(img, (224, 224))

    # 2. Grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 3. Normalization (0–255)
    norm = cv2.normalize(gray, None, alpha=0, beta=255,
                          norm_type=cv2.NORM_MINMAX)

    # 4. Noise Reduction
    blur = cv2.GaussianBlur(norm, (5, 5), 0)

    # 5. Edge Detection
    edges = cv2.Canny(blur, 50, 150)

    return edges

for group, folders in groups.items():
    for folder in folders:
        input_dir = os.path.join(BASE_PATH, group, folder)
        output_dir = os.path.join(OUTPUT_PATH, group, folder)
        os.makedirs(output_dir, exist_ok=True)

        for img_name in os.listdir(input_dir):
            if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
                img_path = os.path.join(input_dir, img_name)
                processed_img = process_image(img_path)

                if processed_img is not None:
                    cv2.imwrite(os.path.join(output_dir, img_name), processed_img)

        print(f"✔ Processed: {group}/{folder}")

print("🎉 All preprocessing steps completed successfully!")
```
